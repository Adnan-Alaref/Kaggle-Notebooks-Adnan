# 🔐 Authenticate with Kaggle API

> **⚠️ Kaggle API Authentication Update**
>
> Kaggle's current CLI provides **four authentication options**:
>
> 1. **OAuth** — `kaggle auth login`
> 2. **Environment Variable** — `KAGGLE_API_TOKEN`
> 3. **API Token File** — store your token in `~/.kaggle/access_token`
> 4. **Legacy API Credentials** — use the old `kaggle.json` file
>
> **In this notebook, we use Option 3: API Token File.**
>
> Therefore, you **do not need to upload `kaggle.json`**. Instead, copy your API token from the Kaggle API settings and store it in:
>
> ```text
> ~/.kaggle/access_token
> ```
>
> 📚 **Official Documentation:**  
> [Kaggle CLI — Authentication](https://github.com/Kaggle/kaggle-cli/tree/main/docs#authentication)

**In Kaggle:**

>**Kaggle → Settings → API → Create New Token**

* **Option 1:**  
  * **Output:**
     * Go to the following link in your browser, and complete the sign-in prompts at Kaggle:

````python
# !kaggle auth login
````

* **Option 3:**  

````python
from google.colab import userdata
KAGGLE_API_TOKEN = userdata.get("KAGGLE_API_TOKEN")
````

````python
import os

os.makedirs(r"/root/.kaggle", exist_ok = True)

with open(r"/root/.kaggle/access_token", mode="w") as file:
  file.write(KAGGLE_API_TOKEN)

os.chmod(r"/root/.kaggle/access_token", 0o600)
````

````python
# !kaggle auth print-access-token
````

**Note:**
>Create `kernel-metadata.json` to tell Kaggle how to import and configure your notebook, including its **title, notebook file, language, GPU, and internet settings**.

````python
!mkdir -p /content/my-notebook
````

````python
!kaggle kernels init -p /content/my-notebook
````

# <a id="Import"></a><div style="background: linear-gradient(to right, #1b5e20, #2e7d32, #388e3c, #43a047, #4caf50); font-family: 'Times New Roman', serif; font-size: 28px; font-weight: bold; text-align: center; border-radius: 15px; padding: 15px; border: 2px solid #ffffff; box-shadow: 0 4px 10px rgba(0, 0, 0, 0.2); -webkit-background-clip: text; -webkit-text-fill-color: transparent;">Step 5: Configure `kernel-metadata.json`.</div>

**For example:**
```json
{
  "id": "user/my-colab-notebook",
  "title": "My Colab Notebook",
  "code_file": "my-notebook.ipynb",
  "language": "python",
  "kernel_type": "notebook",
  "is_private": false,
  "enable_gpu": true,
  "enable_internet": true,
  "dataset_sources": [],
  "competition_sources": [],
  "kernel_sources": [],
  "model_sources": []
}
```
**Important:**   
The `code_file` must exactly match your `.ipynb` filename.
