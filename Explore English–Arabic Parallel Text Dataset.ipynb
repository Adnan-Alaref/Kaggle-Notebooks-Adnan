# English–Arabic Machine Translation with Transformer

This notebook demonstrates how to use the **English–Arabic Parallel Text Dataset** for building an English-to-Arabic machine translation model.

The dataset contains **1,000 parallel English–Arabic sentences** focused on nature and the environment. We will use these sentence pairs to explore the complete machine translation pipeline, from text preprocessing and tokenization to training and evaluating a **Transformer** model in another **NoteBook.**

> **Dataset:** [English–Arabic Parallel Text Dataset](https://www.kaggle.com/datasets/adnanalaref/english-arabic-parallel-text-dataset)

```python
from pathlib import Path

DATA_DIR = Path("/kaggle/input/datasets/adnanalaref/english-arabic-parallel-text-dataset")

english_path = DATA_DIR / "nature_english.txt"
arabic_path = DATA_DIR / "nature_arabic.txt"

with open(english_path, "r", encoding="utf-8") as f:
    english_sentences = [line.strip() for line in f if line.strip()]

with open(arabic_path, "r", encoding="utf-8") as f:
    arabic_sentences = [line.strip() for line in f if line.strip()]

print(f"English sentences: {len(english_sentences)}")
print(f"Arabic sentences:  {len(arabic_sentences)}")

print("\nExample:")
print("English:", english_sentences[0])
print("Arabic: ", arabic_sentences[0])
```

```python
for english, arabic in zip(english_sentences[:5], arabic_sentences[:5]):
    print(f"EN: {english}")
    print(f"AR: {arabic}")
    print("-" * 80)
```
